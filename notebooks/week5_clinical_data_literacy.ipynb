## Vital signs

Vital signs recorded during triage provide an objective assessment of the patient's physiological condition on arrival. These measurements are expected to contribute strongly to emergency department triage decisions.

| Column | Vital | Normal adult Range | Clinical Significance |
|---|---|---|---|
| `triage_vital_hr` | Heart rate (bpm) | 60–100 | > 100 indicates tachycardia, a potential sign of stress or shock. |
| `triage_vital_sbp` | Systolic BP (mmHg) | 90–140 | < 90 suggests possible shock. |
| `triage_vital_dbp` | Diastolic BP (mmHg) | 60–90 | Interpreted alongside SBP for overall blood pressure assessment. |
| `triage_vital_rr` | Respiratory rate (/min) | 12–20 | One of the earliest indicators of clinical deterioration. |
| `triage_vital_o2` | Oxygen saturation (%) | 95–100 | < 92 is concerning, indicating hypoxia. |
| `triage_vital_temp` | Temperature (**°F**) | 97.0–99.5 | Crucially, this is in `Fahrenheit` (not Celsius); fever is typically ≥ 100.4 °F (38 °C). |
| `triage_glucose` | Point-of-care glucose (mg/dL) | 70–140 | Both extremely high and low values are clinically significant. |

Additionally, `triage_vital_o2_device` (coded 0/1) indicates whether supplemental oxygen was in use during SpO₂ measurement, which impacts the interpretation of the oxygen saturation value.

```python
# Define normal adult ranges for key vital signs, noting temperature is in Fahrenheit.
NORMAL_RANGES = {
    "triage_vital_hr":   (60, 100, "bpm"),
    "triage_vital_sbp":  (90, 140, "mmHg"),
    "triage_vital_dbp":  (60,  90, "mmHg"),
    "triage_vital_rr":   (12,  20, "/min"),
    "triage_vital_o2":   (95, 100, "%"),
    "triage_vital_temp": (97.0, 99.5, "F"),   # NOT Celsius
    "triage_glucose":    (70, 140, "mg/dL"),
}
VITALS = list(NORMAL_RANGES)

# Display descriptive statistics for the vital signs.
print(df[VITALS].describe().round(1))

# Count the occurrences of 'triage_vital_o2_device' to see how many patients were on supplemental oxygen.
print("\nOn a supplemental-O2 device?  (0 = room air, 1 = device)")
print(df["triage_vital_o2_device"].value_counts(dropna=False))
```

### Observations

## Chief complaint indicators

Chief complaints are recorded as binary variables indicating whether a particular complaint was documented during triage. Rather than examining each complaint individually, grouping them into broader body systems provides a more meaningful overview of patient presentations.

```python
cc_flags = [c for c in df.columns if c.startswith("cc_")]

# Define categories for chief complaints using keywords. Specific keywords are prioritised.
CC_CATEGORIES = [
    ("Cardiovascular",             ["chestpain","chesttightness","palpitation","irregularheart","rapidheart","tachycard","hypertens","hypotens","cardiacarrest"]),
    ("Genitourinary/Renal",        ["dysuria","hematuria","urinary","flank","testicle","maleguproblem","femaleguproblem","groin","pelvic","vaginal","breastpain","std"]),
    ("Respiratory",                ["breathing","shortnessofbreath","dyspnea","cough","wheez","respiratory","asthma","hemoptysis","coldlike","nasalcongest","influenza","sinus","sorethroat","uri"]),
    ("Neurological",               ["headache","migraine","dizz","syncope","seizure","stroke","alteredmental","confusion","numbness","lossofconsciousness","unresponsive","lethargy","neurologic","hallucinat","extremityweakness"]),
    ("Gastrointestinal",           ["abdominal","epigastric","nausea","emesis","vomit","diarrhea","constipation","gibleeding","giproblem","rectal","swallowedforeign","ingestion","dehydration"]),
    ("Psych/Behavioral/Substance", ["anxiety","depress","agitation","suicid","homicid","psychot","psychiatric","panic","alcohol","drugproblem","drug/alcohol","addiction","detox","overdose","withdrawal","poison"]),
    ("ENT/Eye/Dental",             ["earpain","earproblem","otalgia","eye","conjunctiv","foreignbodyineye","blurredvision","dental","jaw","epistaxis","facial","oralswelling"]),
    ("Skin/Soft-tissue",           ["rash","cellulitis","abscess","cyst","skin","wound","mass","allergicreaction","edema","bruising"]),
    ("Endocrine/Metabolic/Heme",   ["bloodsugar","glycem","glucose","sicklecell"]),
    ("Constitutional/General",     ["fever","chills","fatigue","bodyache","generalizedbody","weakness"]),
    ("Trauma/Injury/MSK",          ["injury","fall","trauma","laceration","fracture","back","neck","joint","rib","assault","burn","animalbite","insectbite","pain","swelling"]),
]

# Function to categorise a chief complaint flag into its body system.
def categorize(flag):
    stem = flag[3:]                     # Remove the "cc_" prefix.
    if "crash" in stem:                 # Handle specific cases like 'crash' before broader terms.
        return "Trauma/Injury/MSK"
    for category, keywords in CC_CATEGORIES:
        if any(k in stem for k in keywords):
            return category
    return "Other/Procedural/Admin"

# Group chief complaint flags by their assigned categories.
from collections import defaultdict
groups = defaultdict(list)
for flag in cc_flags:
    groups[categorize(flag)].append(flag)

# Print the count of flags in each category, ensuring all flags are categorised.
order = [name for name, _ in CC_CATEGORIES] + ["Other/Procedural/Admin"]
for name in order:
    print(f"{name:28s} {len(groups[name]):3d} flags")
print("-" * 40)
print(f"{"TOTAL placed":28s} {sum(len(v) for v in groups.values()):3d} of {len(cc_flags)}")
```

### Observations

Chief complaint indicators account for most of the variables in the dataset. Grouping them by body system provides a clearer summary of patient presentations and highlights which clinical categories occur most frequently.

```python
# (1) Compute the total number of times complaint flags fired within each body system.
print("Complaint volume by body system (flag-hits across all encounters):")
system_totals = {name: int(df[flags].sum().sum()) for name, flags in groups.items()}
# Print system totals, sorted from most to least common.
for name in sorted(system_totals, key=system_totals.get, reverse=True):
    print(f"   {name:28s} {system_totals[name]:>7,}")

# (2) Identify and print the top 5 most common individual complaints.
per_flag = df[cc_flags].sum().sort_values(ascending=False)
print("\nTop 5 individual complaints:")
print(per_flag.head(5))
# Count and print any chief complaint flags that never appeared in the dataset.
print("\nComplaint flags that never fire in this sample:", int((per_flag == 0).sum()))
```

### Observations

The analysis of chief complaints clearly shows that 'Trauma/Injury/MSK', 'Gastrointestinal', and 'Respiratory' are the body systems with the highest reported complaint volumes. Individually, 'abdominalpain', 'other', 'chestpain', and 'shortnessofbreath' are the most frequent complaints. It's also notable that one complaint flag never fired in this dataset, indicating it was not recorded for any patient within this sample.

## Emergency Severity Index (ESI)

The Emergency Severity Index (ESI) represents the triage category assigned to each patient at presentation. As this decision is made during the initial assessment, it is the most appropriate target variable for future predictive modelling.

```python
# Calculate the value counts for ESI and convert to percentages.
counts = df["esi"].value_counts(dropna=False).sort_index()
pct    = (counts / len(df) * 100).round(1)
print("ESI (Triage_Level) distribution:")
# Print the distribution of ESI levels, including counts and percentages.
for level in counts.index:
    print(f"   ESI {level:>3}   {counts[level]:>7,}   ({pct[level]:>4}%)")
# Check for and report any missing ESI labels.
print("\nMissing labels:", int(df["esi"].isna().sum()))
```

### Observations

The ESI variable contains no missing values and therefore provides a complete target for supervised learning. In contrast, disposition represents a later clinical outcome and should not be used as an input feature because it would introduce data leakage.

## ESI Severity Levels

The ESI system classifies patients into five levels according to the urgency of their condition. Lower scores represent higher clinical acuity, while higher scores represent less urgent presentations.

```python
# Define a mapping from ESI numerical levels to their descriptive names.
labels = {1: "Resuscitation", 2: "Emergent", 3: "Urgent", 4: "Less urgent", 5: "Non-urgent"}
# Get the counts of each ESI level and sort by index.
counts = df["esi"].value_counts().sort_index()
print("ESI  ·  name           ·  encounters")
# Print each ESI level with its name and the number of encounters.
for level, n in counts.items():
    print(f"   {int(level)}  ·  {labels[int(level)]:<13}  ·  {n:>7,}")
```

### Observations

Most patients fall within the middle acuity levels, while the highest priority cases are comparatively uncommon. This distribution is consistent with typical emergency department activity.
